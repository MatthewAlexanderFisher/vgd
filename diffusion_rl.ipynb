# RL for Adaptive Time-Changed Markov Processes

## Continuous-Time Markov Processes

### 1. Setting

Given:

* A measurable state space $(E, \mathcal{E})$, usually Polish;
* A càdlàg process $Y = (Y_t)_{t \ge 0}$ taking values in $E$;
* A filtered probability space $(\Omega, \mathcal{F}, (\mathcal{F}_t)_{t \ge 0}, \mathbb{P})$ satisfying the usual conditions.

Goal: specify a **time-homogeneous Markov process**.

### 2. Transition Structure

For all $s,t \ge 0$ and $A \in \mathcal{E}$,
$$
\mathbb{P}(Y_{t+s} \in A \mid \mathcal{F}_t) = P_s(Y_t, A),
$$
where $(P_t)_{t \ge 0}$ is a **semigroup of Markov kernels** satisfying
$$
P_{t+s}(x,A) = \int_E P_t(x,dy)\,P_s(y,A),
$$
and acting on functions via
$$
T_t f(x) = \int_E f(y)\,P_t(x,dy).
$$

The initial distribution $\mu$ on $(E, \mathcal{E})$ satisfies $Y_0 \sim \mu$.

### 3. Weak (Extended) Generator

The **weak generator** $\tilde{\mathcal{L}}$ is defined by the **martingale problem**:

A measurable $f : E \to \mathbb{R}$ belongs to $\mathrm{Dom}(\tilde{\mathcal{L}})$ if there exists $g : E \to \mathbb{R}$ such that
$$
M_t^f := f(Y_t) - f(Y_0) - \int_0^t g(Y_s)\,ds
$$
is a martingale for all initial conditions.  
In this case, write $g = \tilde{\mathcal{L}} f$.

This generalises the strong (pointwise) generator
$$
\mathcal{L} f = \lim_{t \downarrow 0} \frac{T_t f - f}{t},
$$
and applies to general càdlàg (not necessarily Feller) processes.

**Examples:**
* Countable $E$: $\tilde{\mathcal{L}}$ is the rate matrix $Q$.
* Diffusions on $E \subseteq \mathbb{R}^d$:
  $$
  (\tilde{\mathcal{L}} f)(x) = b(x)^\top \nabla f(x)
  + \tfrac{1}{2}\operatorname{Tr}\big(\sigma(x)\sigma(x)^\top \nabla^2 f(x)\big).
  $$

### 4. Canonical Construction

Let
$$
D_E[0,\infty) = \{\omega : [0,\infty) \to E \text{ càdlàg}\}, \quad
Y_t(\omega) = \omega(t), \quad
\mathcal{F}_t = \sigma(Y_s : s \le t).
$$
Given an initial law $\mu$ and transition semigroup $(P_t)$,  
the **Kolmogorov extension theorem** yields a unique measure $\mathbb{P}^\mu$ on $D_E[0,\infty)$ such that
$$
\mathbb{P}^\mu(Y_{t_1} \in A_1, \ldots, Y_{t_n} \in A_n)
= \int_E \mu(dx_0)\, P_{t_1-t_0}(x_0,dx_1)\cdots P_{t_n-t_{n-1}}(x_{n-1},A_n).
$$


### 5. Minimal Specification

| Ingredient              | Symbol                          | Role                                                  |
| ----------------------- | ------------------------------- | ----------------------------------------------------- |
| State space             | $(E,\mathcal{E})$               | Possible states                                       |
| Transition semigroup    | $(P_t)_{t \ge 0}$               | Defines Markov/time-homogeneous evolution             |
| Initial distribution    | $\mu$                           | Law of $Y_0$                                          |
| Weak generator          | $\tilde{\mathcal{L}}$                      | Infinitesimal characterisation via martingale problem |
| Path space & filtration | $(D_E[0,\infty), (\mathcal{F}_t))$ | Measure-theoretic setting                             |



## Time-Changed Markov Processes

A **time-changed** Markov process is a stochastic process where the evolution of the process is governed by a Markov process, but the time parameter is modified:

$$
X_t = Y_{r(t)}, \quad r(t) = \int_0^t s(X_u) \mathrm{d}u.
$$

Here, $Y_t$ is the **base** Markov process, and $r(t)$ is a time-change function that depends on the state of the process through a **speed function** $s(x)$. The function $s(x)$ determines how quickly or slowly time progresses based on the current state $X_u$.

## Continuous-Time Markov Decision Processes

A **continuous-time Markov decision process** ([CTMDP](https://books.google.co.uk/books/about/Continuous_Time_Markov_Decision_Processe.html?id=tgi-opMlLTwC&hl=en&redir_esc=y)) is a generalisation of a continuous-time Markov process that incorporates decision-making capabilities. 

The **general ingredients** are:

1. **State space:** A measurable space $(E, \mathcal{E})$, often a Polish space.

2. **Action space:** A measurable space $(A, \mathcal{A})$ with feasible set $A(x) \subseteq A$ for each $x \in E$.  
   

3. **Probability set-up:** A filtered probability space $(\Omega, \mathcal{F}, (\mathcal{F}_t)_{t \ge 0}, \mathbb{P})$ satisfying the usual conditions, with initial distribution $Y_0 \sim \mu_0$ (or fixed $x_0$).

4. **Admissible controls:** Processes $(a_t)_{t \ge 0}$ such that:
   * **Values:** $a_t \in A(Y_t)$ a.s.;
   * **Non-anticipativity:** $(a_t)$ is $(\mathcal{F}_t)$-progressively (or predictably) measurable;
   * **Control classes:**
     - *Markov/feedback:* $a_t = \pi(Y_t)$;
     - *Stationary Markov:* $a_t = \pi(Y_t)$ with $\pi$ time-independent;
     - *Relaxed:* $a_t$ is a probability measure on $A$ 


Define a **family of (weak) generators** $\{\mathcal{L}^a : a \in A\}$ with common core $\mathcal{D} \subseteq C_b(E)$ such that, for any admissible control $a_\cdot$, the process $Y$ satisfies the **controlled martingale problem**:

$$
M_t^f := f(Y_t) - f(Y_0) - \int_0^t (\mathcal{L}^{a_s} f)(Y_s) \, ds
\quad \text{is a martingale for all } f \in \mathcal{D}.
$$

This condition uniquely determines the law of $Y$ under any control.


### Constraints for the Sampling Objective

In the MCMC/sampling context, we additionally require generators $\mathcal{L}^a$ whose **invariant distribution** (or stationary measure) is the target $\mu$.

That means:
$$
\int_E \mathcal{L}^a f(x) \, \pi(dx) = 0 \quad \text{for all suitable } f.
$$

Equivalently, if $\mathcal{L}^*$ is the adjoint of $\mathcal{L}$ on $\mathcal{L}^2(\pi)$, then:
$$
\mathcal{L}^* \pi = 0.
$$
That is, under the dynamics defined by $\mathcal{L}$, the probability mass of $\mu$ doesn’t change in time:
$$
\frac{d}{dt}\int_E f(x)\,\pi(dx) = 0.
$$


### Time-Changed Markov Processes Controller Model

Let $\tilde{\mathcal{L}}$ be the (weak) generator of the base process $Y_t$. By Proposition B.1 in [Sampling for Time-Changed Markov Processes](https://arxiv.org/pdf/2501.15155v1), the time-changed process $X_t = Y_{r(t)}$ with $r(t) = \int_0^t s(X_u) du$ then the weak generator $\mathcal{L}$ of $X_t$ satisfies, on a suitable core,
$$
\mathcal{L} f(x)= s(x)\,\tilde{\mathcal{L}} f(x).
$$
So if we parameterise the speed as $s_a(\cdot)$ by an action $a$ (or policy parameters), we obtain a **controlled family**
$$
\mathcal{L}^{a} f(x)= s_{a}(x) \, \tilde{\mathcal{L}} f(x),
$$
i.e. a continuous-time MDP where the control acts by **multiplicative time rescaling** of the base dynamics. 

### Value Function and Objective

Given a reward function $\rho:E\times A\rightarrow \mathbb{R}$, we define the **value function** for a policy $\pi$ as

$$
V^\pi(x) = \mathbb{E}^\pi_x\left[\int_0^\infty e^{-\gamma t} \rho(X_t, \pi(X_t)) \, dt\right].
$$

It satisfies the **Hamilton-Jacobi-Bellman (HJB) equation**:

$$
\gamma V^\pi(x) = \sup_{a \in A(x)} \left\{ \rho(x,a) + (\mathcal{L}^a V^\pi)(x) \right\}.
$$

#### Rough Derivation

Follows from Dynkin's formula/Martingale problem:

$$M_t^f = f(X_t) - f(X_0) - \int_0^t (\mathcal L f)(X_s)\,ds$$

### Target Distributions

Targets:

$$
\pi(x) \propto e^{-U(x)}
$$

with $U(x)$ the potential function and $\nabla U(x)$ its gradient.



```python
from jax.scipy.special import logsumexp
import jax.numpy as jnp
import jax

# Gaussian density
def make_gaussian_density(mean, cov):
    mean = jnp.asarray(mean)
    cov  = jnp.asarray(cov)

    d = mean.shape[0]
    inv = jnp.linalg.inv(cov)
    sign, logdet = jnp.linalg.slogdet(cov)
    assert sign > 0, "Covariance must be SPD."

    log_norm = -0.5 * (d * jnp.log(2.0 * jnp.pi) + logdet)

    @jax.jit
    def logpdf(x):
        x = jnp.asarray(x)
        diff = x - mean
        # quadratic form (...,)
        quad = jnp.einsum('...i,ij,...j->...', diff, inv, diff)
        return log_norm - 0.5 * quad

    @jax.jit
    def pdf(x):
        return jnp.exp(logpdf(x))

    return pdf, logpdf


# Gaussian grad U
def make_gaussian_grad_potential(mean, cov):
    inv_cov = jnp.linalg.inv(cov)
    def grad_U(x):
        return inv_cov @ (x - mean)
    return jax.jit(grad_U)


# GMM grad U
def make_gmm_grad_potential(means, covs, weights):
    """
    means:   (K, d)
    covs:    (K, d, d)   
    weights: (K,)        (positive, will be normalised)
    returns: grad_U(x) function mapping (d,) -> (d,)
    """
    means   = jnp.asarray(means)
    covs    = jnp.asarray(covs)
    weights = jnp.asarray(weights)
    weights = weights / jnp.sum(weights)

    K, d = means.shape

    # Precompute precisions and log-normaliser terms
    precisions = jnp.linalg.inv(covs)                        # (K,d,d)
    sign, logdet = jnp.linalg.slogdet(covs)                 # (K,)
    assert_val = jnp.all(sign > 0)                          # for sanity in eager mode
    # log of w_k * normaliser of N(x|mu, Sigma) without quad term
    log_const = jnp.log(weights) - 0.5 * (d * jnp.log(2.0 * jnp.pi) + logdet)  # (K,)

    @jax.jit
    def grad_U(x):
        """
        Gradient of potential: ∇U(x) = -∇ log π(x).
        """
        x = jnp.asarray(x)
        diff = x[None, :] - means                            # (K,d)
        quad = jnp.einsum('ki,kij,kj->k', diff, precisions, diff)  # (K,)
        log_comp = log_const - 0.5 * quad                    # (K,)
        log_pi = logsumexp(log_comp)        # scalar
        gamma = jnp.exp(log_comp - log_pi)                   # (K,)
        grad = jnp.einsum('k,kij,kj->j', gamma, precisions, diff)  # (d,)
        return grad

    return grad_U

# GMM density
def make_gmm_density(means, covs, weights):
    """
    means:   (K, d)
    covs:    (K, d, d) SPD
    weights: (K,) nonnegative (will be normalised)
    returns: (pdf, logpdf) where each maps (..., d) -> (...)
    """
    means   = jnp.asarray(means)
    covs    = jnp.asarray(covs)
    weights = jnp.asarray(weights)
    weights = weights / jnp.sum(weights)

    K, d = means.shape

    precisions = jnp.linalg.inv(covs)                         # (K,d,d)
    sign, logdet = jnp.linalg.slogdet(covs)                  # (K,)
    assert jnp.all(sign > 0), "All covariances must be SPD."

    # log w_k minus the normalising constants of each component
    log_consts = (jnp.log(weights)
                  - 0.5 * (d * jnp.log(2.0 * jnp.pi) + logdet))  # (K,)

    @jax.jit
    def comp_logpdfs(x):
        """
        x: (..., d)
        returns: (..., K) with log N_k(x) + log w_k
        """
        x = jnp.asarray(x)
        # diff: (..., K, d)
        diff = x[..., None, :] - means[None, ...]
        # quad: (..., K)
        quad = jnp.einsum('...kd,kdj,...kj->...k', diff, precisions, diff)
        return log_consts[None, ...] - 0.5 * quad

    @jax.jit
    def logpdf(x):
        # log-sum-exp across components
        return logsumexp(comp_logpdfs(x), axis=-1)

    @jax.jit
    def pdf(x):
        return jnp.exp(logpdf(x))

    return pdf, logpdf
```

### Zig-Zag Sampler

From [this paper](https://arxiv.org/pdf/1607.03188), the Zig-Zag process is a piecewise deterministic Markov process (PDMP) on $\mathbb{R}^d \times \{-1,+1\}^d$ with state $(X_t, V_t)$, where $X_t \in \mathbb{R}^d$ is the position and $V_t \in \{-1,+1\}^d$ is the velocity.

The rate is state-dependent which yields an inhomogeneous Poisson process for the switching times:

$$
\lambda_i(x,v) = \big(v_i \partial_{x_i} U(x)\big)_+.
$$

To simulate the Zig-Zag process, we require a method to sample the switching times from the inhomogeneous Poisson process with rate $\lambda_i(x,v)$. There are three approaches:

1. **Direct sampling**: If the integrated rate function $\Lambda_i(t; x, v) = \int_0^t \lambda_i(x + vs, v) ds$ has a closed-form inverse, we can sample the next switching time $T_i$ by drawing $E \sim \text{Exponential}(1)$ and setting $T_i = \Lambda_i^{-1}(E; x, v)$.
2. **Thinning**: If $\lambda_i(x + vs, v)$ is Lipschitz for all $s \geq 0$, we can simulate a inhomogeneous Poisson process with linear rate $\bar{\lambda}_i$ and accept each event with probability $\lambda_i(x + vs, v) / \bar{\lambda}_i(x + vs, v)$.
3. **(Strang) Splitting + Metropolis-Adjustment**: I.e.[Algorithm 3](https://arxiv.org/pdf/2301.02537). Roughly: decompose the generator of the Zig-Zag process into a sum of simpler generators, simulate each component separately, and combine them using a Metropolis-Hastings acceptance step to ensure the correct invariant distribution.

```python
import jax.numpy as jnp
import jax
from matplotlib import pyplot as plt
import numpy as np

from vgd.sampler.zigzag import run_zigzag

K, d = 13, 2
means = jnp.array([[ -20.0, 20.0],
                   [  -0.,  20.0],
                   [20.0,  20.0],
                   [-10.0,   10.],
                   [10.0,   10.],
                   [  -20.,    0.],
                   [   0.,    0.],
                   [  20.,    0.],
                   [ -10.,  -10.],
                   [  10.,  -10.],
                   [   -0.,  -20.],
                   [  -20., -20.],
                   [  20., -20.]
                   ])
covs = jnp.stack([jnp.eye(d) * 2.0 for _ in range(K)
])
weights = jnp.ones((K,)) / K

target_grad_potential = make_gmm_grad_potential(means, covs, weights)
target_pdf = make_gmm_density(means, covs, weights)[0]


# Run sampling
d = 2
key = jax.random.PRNGKey(42)
x0 = jnp.zeros((d,))
key, k_v = jax.random.split(key)
v0 = jnp.where(jax.random.bernoulli(k_v, 0.5, (d,)), 1.0, -1.0)
L  = jnp.ones((d,))

print("Running Zig-Zag sampler...")
xs, vs, ts, accepted, which, accp = zigzag_sample(
    key, target_grad_potential, x0, v0, L, num_events=10_000, refresh_rate=0.0
)

# Calculate statistics
burn_in = 200
xs_samples = xs[burn_in:]
acceptance_rate = jnp.mean(accepted)
mean_estimate = jnp.mean(xs_samples, axis=0)
cov_estimate = jnp.cov(xs_samples.T)

print(f"\nAcceptance rate: {acceptance_rate:.3f}")
print(f"Mean estimate: {mean_estimate}")
print(f"Covariance estimate:\n{cov_estimate}")
print(f"Total trajectory time: {ts[-1]:.2f}")

# Visualisation

fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# Plot 1: Trajectory
ax = axes[0, 0]
NN = 9000
ax.plot(xs_samples[:NN, 0], xs_samples[:NN, 1], 'b-', alpha=0.3, linewidth=0.5)
ax.scatter(xs_samples[:NN, 0], xs_samples[:NN, 1], c=range(NN), cmap='viridis', s=1, alpha=0.6)
ax.set_xlabel('$x_1$')
ax.set_ylabel('$x_2$')
ax.set_title('Zig-Zag Trajectory')
ax.grid(True, alpha=0.3)
# ax.axis('equal')

# Plot 2: Sample distribution vs target
ax = axes[0, 1]
min_val, max_val = -30, 30
# Target contours
x_range = np.linspace(min_val, max_val, 300)
y_range = np.linspace(min_val, max_val, 300)
X, Y = np.meshgrid(x_range, y_range)
Z = target_pdf(np.stack([X, Y], axis=-1))
ax.contourf(X, Y, Z, levels=10, cmap='viridis')
# Samples
ax.scatter(np.array(xs_samples[:, 0]), np.array(xs_samples[:, 1]), c='red', s=2, alpha=0.1)
ax.set_xlabel('$x_1$')
ax.set_ylabel('$x_2$')
ax.set_title('Sample Distribution vs Target')
# ax.axis('equal')
ax.set_xlim([min_val, max_val])
ax.set_ylim([min_val, max_val])

# Plot 3: Marginal distributions
ax = axes[1, 0]
ax.hist(np.array(xs_samples[:, 0]), bins=50, density=True, alpha=0.7, label='x₁ samples')
ax.hist(np.array(xs_samples[:, 1]), bins=50, density=True, alpha=0.7, label='x₂ samples')
x_plot = np.linspace(-4, 4, 200)
# ax.plot(x_plot, np.exp(-0.5 * x_plot**2) / np.sqrt(2 * np.pi), 'r-', linewidth=2, label='N(0,1)')
ax.set_xlabel('Value')
ax.set_ylabel('Density')
ax.set_title('Marginal Distributions')
ax.legend()
ax.grid(True, alpha=0.3)

# Plot 4: Trace plots
ax = axes[1, 1]
n_plot = min(NN, len(xs_samples))
ax.plot(xs_samples[:n_plot, 0], alpha=0.7, label='x₁')
ax.plot(xs_samples[:n_plot, 1], alpha=0.7, label='x₂')
ax.set_xlabel('Iteration')
ax.set_ylabel('Value')
ax.set_title('Trace Plots')
ax.legend()
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

#### (Time Changed) Metropolis-Adjusted Zig-Zag (MAZZ) Sampler

Uses splitting + Metropolis-adjustment. I.e. decomposes:

$$ \mathcal{L} f(x,v)
= \langle v, \nabla_x f(x,v)\rangle + \sum_{i=1}^d \lambda_i(x,v)\,[f(x, R_i v) - f(x,v)].$$

into **drift** and **jump** (switching) components (recall $R_i$ flips the $i$-th coordinate of $v$).

However, this **discretises time** (unlike the original Zig-Zag which is event-driven).

```python
import jax.numpy as jnp
import jax
from matplotlib import pyplot as plt
import numpy as np

from vgd.sampler.zigzag import run_mazz

K, d = 13, 2
means = jnp.array([[ -20.0, 20.0],
                   [  -0.,  20.0],
                   [20.0,  20.0],
                   [-10.0,   10.],
                   [10.0,   10.],
                   [  -20.,    0.],
                   [   0.,    0.],
                   [  20.,    0.],
                   [ -10.,  -10.],
                   [  10.,  -10.],
                   [   -0.,  -20.],
                   [  -20., -20.],
                   [  20., -20.]
                   ])
covs = jnp.stack([jnp.eye(d) * 3.0 for _ in range(K)
])
weights = jnp.ones((K,)) / K

target_grad_potential = make_gmm_grad_potential(means, covs, weights)
target_pdf, target_log_pdf = make_gmm_density(means, covs, weights)

alpha = 0.9
beta  = 1e-5
log_beta = jnp.log(beta)

def U(x):
    return -target_log_pdf(x)  # you already have this

def speed_fn(x):
    # s(x) = (π + β)^(-α)  with  π = exp(-U)
    logpi   = -U(x)                                  # = target_log_pdf(x)
    logsum  = jnp.logaddexp(logpi, log_beta)         # log(π + β)
    return jnp.exp(-alpha * logsum)                  # (π+β)^(-α)

def grad_speed_fn(x):
    # ∇s = α * w * s * ∇U,  where  w = π / (π + β)
    logpi   = -U(x)
    logsum  = jnp.logaddexp(logpi, log_beta)         # log(π + β)
    s       = jnp.exp(-alpha * logsum)
    w       = jnp.exp(logpi - logsum)                # π / (π + β) in log-space
    return alpha * w * s * target_grad_potential(x)


# Run sampling
d = 2
key = jax.random.PRNGKey(42)
x0 = jnp.zeros((d,))
key, k_v = jax.random.split(key)
v0 = jnp.where(jax.random.bernoulli(k_v, 0.5, (d,)), 1.0, -1.0)
L  = jnp.ones((d,))
num_steps = 40_000

print("Running (Time Changed) MAZZ sampler...")
xs, vs, stats = run_mazz(
    key, x0, v0, delta=1e-3, num_steps=num_steps,
    grad_potential=target_grad_potential,
    speed_fn=speed_fn,
    grad_speed_fn=grad_speed_fn,
    potential=U,       
)
```

#### Time-Changed Zig-Zag (TCZZ) Sampler

If we assume a speed function of the form

$$
s(x) = (\pi(x) + \beta)^{-\alpha} \quad \text{for } \alpha, \beta > 0,
$$

then, under constraints on the target $\pi$, the rate function

$$
\lambda_i(x,v) = \,(v_i (s(x))\partial_{x_i} U(x) - \partial_{x_i} s(x))_+ + s(x)\gamma_i(x)
$$

is **locally Lipschitz** and can be upper bounded by a linear function, allowing the use of the **thinning** method to sample switching times.




With Prop. 3.4 and $s = s(x)$ (no $v$ dependence), the deterministic flow is  
$$
\dot x(t) = s(x(t))\,v, \qquad v \in \{\pm1\}^d,
$$
and the **core** (pre-ReLU) rate for coordinate $i$ is  
$$
c_i(x,v) := v_i\big(s(x)\,\partial_i U(x) - \partial_i s(x)\big).
$$
Then  
$$
\lambda_i(x,v) = [\,c_i(x,v)\,]_+ + s(x)\,\gamma\, e_i(x).
$$

To build a linear envelope for $\lambda_i(x(t),v)$ along the path, it suffices to control the **time derivative** of $c_i$ along the trajectory:
$$
\frac{d}{dt} c_i(x(t),v)
= \nabla c_i(x)^\top \dot x
= \nabla c_i(x)^\top \big(s(x)\,v\big).
$$
If $\big|\tfrac{d}{dt} c_i\big| \le B_i$ on the current segment, then
$$
\lambda_i(t) \le 
\underbrace{[c_i(x_0,v)]_+}_{a_i} + 
\underbrace{B_i}_{b_i}\,t 
\quad\Rightarrow\quad 
\bar\lambda_i(t) = a_i + b_i t
$$
is a valid **linear dominating rate** for thinning on that segment.

For $s(x) = (\pi(x) + \beta)^{-\alpha}$ you can compute $\nabla c_i$ cheaply

Write $U = -\log \pi$, $\pi = e^{-U}$, and
$$
s(x) = (\pi + \beta)^{-\alpha} = e^{-\alpha \log(\pi + \beta)}.
$$
Define $w(x) = \dfrac{\pi}{\pi + \beta} \in (0,1)$.  
Then (derived earlier)
$$
\nabla s(x) = \alpha\,w(x)\,s(x)\,\nabla U(x), \qquad
\partial_i s(x) = \alpha\,w(x)\,s(x)\,\partial_i U(x).
$$
Hence the core simplifies to
$$
c_i(x,v) = v_i\big(1 - \alpha w(x)\big)\,s(x)\,\partial_i U(x).
$$
Differentiate:
$$
\nabla c_i(x,v)
= \underbrace{\nabla\!\big[(1 - \alpha w)s\big]}_{\text{scalar grad}}\;\partial_i U
+ (1 - \alpha w)\,s\,\underbrace{\nabla(\partial_i U)}_{H_U[:,i]}.
$$
A short calculus exercise (using $\nabla w = -w(1 - w)\,\nabla U$) gives
$$
\nabla\!\big[(1 - \alpha w)s\big]
= s\,\alpha\,w\,\big(2 - (1 + \alpha)\,w\big)\,\nabla U.
$$
Therefore
$$
\nabla c_i(x,v)
= s\,\alpha\,w\,\big(2 - (1 + \alpha)\,w\big)\,(\partial_i U)\,\nabla U
+ (1 - \alpha w)\,s\,H_U[:,i].
$$

Now along the path $\dot x = s\,v$,
$$
\frac{d}{dt} c_i(x(t),v)
= \nabla c_i^\top (s v)
= s^2\Big[
\alpha w\big(2 - (1 + \alpha)w\big)\,(\partial_i U)\,(\nabla U^\top v)
+ (1 - \alpha w)\,(H_U v)_i
\Big].
$$
